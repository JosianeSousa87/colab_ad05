# 🐶 Versão 1: uma imagem de cachorro

Faz uma requisição à API [Dog CEO](https://dog.ceo/dog-api/), lê o endereço da imagem na resposta JSON, baixa e salva em `imagens/dog/dog.jpg`.

> No Colab os arquivos ficam em `/content` e **são apagados quando a sessão termina**. Baixe o que quiser guardar pelo painel de arquivos (ícone de pasta, à esquerda).

## 1. Importações

```python
import os
import requests
from IPython.display import Image, display
```

## 2. Pasta de saída

```python
pasta = os.path.join(os.getcwd(), 'imagens/dog')
os.makedirs(pasta, exist_ok=True)
caminho = os.path.join(pasta, 'dog.jpg')
```

## 3. Requisição à API e download da imagem

```python
url = 'https://dog.ceo/api/breeds/image/random'
response = requests.get(url)
data = response.json()
img_url = data['message']
img_data = requests.get(img_url).content
with open(caminho, 'wb') as handler:
    handler.write(img_data)
print(f"Imagem do cachorro salva em {caminho}!")
```

## 4. Visualizar a imagem

```python
display(Image(filename=caminho, width=400))
```
